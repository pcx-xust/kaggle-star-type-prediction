```python
# ============================================================
# Overall model comparison: baseline vs color features
# ============================================================

import pandas as pd
from pathlib import Path

# 如果前面已经定义了PROJECT_DIR，这里会直接使用；
# 如果没有定义，则自动从当前目录向上查找项目根目录。
try:
    PROJECT_DIR
except NameError:
    CURRENT_DIR = Path.cwd()
    PROJECT_DIR = None

    for path in [CURRENT_DIR] + list(CURRENT_DIR.parents):
        has_data = (path / "data").exists()
        has_notebook_dir = (path / "notebook").exists() or (path / "notebooks").exists()

        if has_data and has_notebook_dir:
            PROJECT_DIR = path
            break

    if PROJECT_DIR is None:
        raise FileNotFoundError(
            "Cannot locate project root. Please check the current working directory."
        )

# 总模型结果保存目录
OVERALL_MODEL_DIR = PROJECT_DIR / "reports" / "model"
OVERALL_MODEL_DIR.mkdir(parents=True, exist_ok=True)

# 汇总所有阶段模型结果
overall_model_comparison = pd.DataFrame({
    "Stage": [
        "Baseline",
        "Baseline",
        "Baseline",
        "Color Features",
        "Color Features"
    ],
    "Model": [
        "Dummy",
        "RandomForest",
        "ExtraTrees",
        "RandomForest",
        "ExtraTrees"
    ],
    "Accuracy": [
        0.653815,
        0.956370,
        0.956188,
        0.959470,
        0.954949
    ],
    "Balanced Accuracy": [
        0.333333,
        0.953654,
        0.932845,
        0.942999,
        0.930951
    ],
    "Macro F1": [
        0.263558,
        0.942153,
        0.938546,
        0.944721,
        0.936589
    ]
})

# 按Macro F1从高到低排序
overall_model_comparison = overall_model_comparison.sort_values(
    by="Macro F1",
    ascending=False
).reset_index(drop=True)

# 保存CSV文件
overall_model_comparison.to_csv(
    OVERALL_MODEL_DIR / "overall_model_comparison.csv",
    index=False,
    encoding="utf-8-sig"
)

# 手动生成Markdown表格，避免依赖tabulate包
markdown_lines = []
markdown_lines.append("| Stage | Model | Accuracy | Balanced Accuracy | Macro F1 |")
markdown_lines.append("|---|---|---:|---:|---:|")

for _, row in overall_model_comparison.iterrows():
    markdown_lines.append(
        f"| {row['Stage']} | "
        f"{row['Model']} | "
        f"{row['Accuracy']:.6f} | "
        f"{row['Balanced Accuracy']:.6f} | "
        f"{row['Macro F1']:.6f} |"
    )

markdown_table = "\n".join(markdown_lines)

with open(
    OVERALL_MODEL_DIR / "overall_model_comparison.md",
    "w",
    encoding="utf-8"
) as f:
    f.write(markdown_table)

# 显示总比较表
overall_model_comparison
```
